**3. Discussions:**

A data dashboard is a visual interface that displays key information and performance indicators in a concise and interactive manner. Dashboards consolidate multiple data sources to provide users with insights for decision-making and performance tracking.

Principles of Effective Dashboards:

Clarity and Simplicity: Visuals should be easy to read and interpret.

Real-Time or Dynamic Data: Dashboards often pull updated data automatically.

Interactivity: Users should be able to filter, sort, or drill down into specific metrics.

Action-Oriented Insights: The dashboard should guide users toward decisions or next steps.

Taxonomies of Data Dashboards:

Data Updates – Dashboards can show static (snapshot) or dynamic (real-time) data.

User Interaction – Some dashboards allow filtering, searching, and selecting parameters.

Organizational Function – Dashboards can serve strategic, analytical, or operational functions.

For example:

Strategic dashboards help top management monitor KPIs.

Analytical dashboards help analysts explore data trends.

Operational dashboards support real-time monitoring of systems and workflows.

```python
import pandas as pd
import numpy as np
import plotly.express as px
from plotly.subplots import make_subplots
import plotly.graph_objects as go
```

```python
# Simulated company performance data
np.random.seed(42)
months = pd.date_range("2024-01-01", periods=12, freq="M")
regions = ["North", "South", "East", "West"]
products = ["Laptop", "Tablet", "Phone"]

data = {
    "Month": np.random.choice(months, 200),
    "Region": np.random.choice(regions, 200),
    "Product": np.random.choice(products, 200),
    "Sales": np.random.randint(5000, 20000, 200),
    "Profit": np.random.randint(500, 5000, 200)
}

df = pd.DataFrame(data)
df.head()
```

```python
# Aggregated views
sales_by_region = df.groupby("Region", as_index=False)["Sales"].sum()
sales_by_product = df.groupby("Product", as_index=False)["Sales"].sum()
sales_trend = df.groupby("Month", as_index=False)["Sales"].sum().sort_values("Month")

# Create dashboard layout
fig = make_subplots(rows=2, cols=2,
                    subplot_titles=("Sales by Region", "Sales by Product", "Monthly Sales Trend", "Profit vs Sales"),
                    specs=[[{"type": "bar"}, {"type": "pie"}],
                           [{"colspan": 2}, None]])

# Add charts
fig.add_trace(go.Bar(x=sales_by_region["Region"], y=sales_by_region["Sales"], name="Region Sales"), row=1, col=1)
fig.add_trace(go.Pie(labels=sales_by_product["Product"], values=sales_by_product["Sales"], name="Product Sales"), row=1, col=2)
fig.add_trace(go.Scatter(x=sales_trend["Month"], y=sales_trend["Sales"], mode="lines+markers", name="Monthly Trend"), row=2, col=1)

fig.update_layout(height=700, width=950, title_text="📊 Interactive Data Dashboard Example")
fig.show()
```

```python
# Interactive dashboard: filterable by product
fig = px.bar(df, x="Region", y="Sales", color="Product", barmode="group",
             title="Interactive Dashboard: Sales by Region and Product",
             hover_data=["Profit"])
fig.show()
```

```python
#Code Here
#Student Performance Dashboard

#Student performance data
np.random.seed(42)
students = [f"Student {i}" for i in range(1, 201)]
subjects = ["Math", "Science", "English", "History"]
regions = ["North", "South", "East", "West"]
genders = ["Male", "Female"]

data = {
    "Student": np.random.choice(students, 200),
    "Region": np.random.choice(regions, 200),
    "Gender": np.random.choice(genders, 200),
    "Subject": np.random.choice(subjects, 200),
    "Score": np.random.randint(50, 100, 200)
}

df = pd.DataFrame(data)

# Aggregated views
avg_score_by_subject = df.groupby("Subject", as_index=False)["Score"].mean()
avg_score_by_region = df.groupby("Region", as_index=False)["Score"].mean()
avg_score_by_gender = df.groupby("Gender", as_index=False)["Score"].mean()
avg_score_by_subject_gender = df.groupby(["Subject", "Gender"], as_index=False)["Score"].mean()

# --- DASHBOARD LAYOUT ---
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=(
        "Average Score by Subject",
        "Average Score by Region",
        "Average Score by Gender",
        "Subject Performance by Gender"
    ),
    specs=[
        [{"type": "bar"}, {"type": "bar"}],
        [{"type": "pie"}, {"type": "bar"}]
    ]
)

# Chart 1: Average Score by Subject
fig.add_trace(
    go.Bar(x=avg_score_by_subject["Subject"], y=avg_score_by_subject["Score"],
           marker_color='dodgerblue', name="Subject Average"),
    row=1, col=1
)

# Chart 2: Average Score by Region
fig.add_trace(
    go.Bar(x=avg_score_by_region["Region"], y=avg_score_by_region["Score"],
           marker_color='mediumseagreen', name="Regional Average"),
    row=1, col=2
)

# Chart 3: Average Score by Gender (Pie)
fig.add_trace(
    go.Pie(labels=avg_score_by_gender["Gender"], values=avg_score_by_gender["Score"],
           name="Gender Average", hole=0.4),
    row=2, col=1
)

# Chart 4: Subject Performance by Gender (Grouped Bar)
for gender in avg_score_by_subject_gender["Gender"].unique():
    sub_df = avg_score_by_subject_gender[avg_score_by_subject_gender["Gender"] == gender]
    fig.add_trace(
        go.Bar(x=sub_df["Subject"], y=sub_df["Score"], name=gender),
        row=2, col=2
    )

# --- STYLE & LAYOUT ---
fig.update_layout(
    title_text="Student Performance Dashboard",
    height=750,
    width=1000,
    showlegend=True,
    legend_title="Gender",
    template="plotly_white"
)

fig.update_xaxes(title_text="Category")
fig.update_yaxes(title_text="Average Score")

fig.show()

# --- OPTIONAL INTERACTIVE DASHBOARD ---
fig2 = px.bar(
    df,
    x="Subject",
    y="Score",
    color="Gender",
    barmode="group",
    facet_col="Region",
    title="Interactive Dashboard: Student Scores by Subject, Gender, and Region",
    hover_data=["Student"]
)
fig2.update_layout(height=700, width=1000, template="simple_white")
fig2.show()
```

**Task 2:** Data Dashboard Taxonomies

Add interactivity features such as:

A dropdown menu to select regions or products

A dynamic update (simulating real-time data) using Python’s update functions

Explain how your dashboard fits into:

Data Update Type (Static, Scheduled, or Real-Time)

User Interaction (Filter, Drilldown, etc.)

Organizational Function (Strategic, Analytical, Operational)

```python
#Student Performance Dashboard with Working Dropdown Filter
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

#Simulated data
np.random.seed(42)
students = [f"Student {i}" for i in range(1, 201)]
subjects = ["Math", "Science", "English", "History"]
regions = ["North", "South", "East", "West"]
genders = ["Male", "Female"]

data = {
    "Student": np.random.choice(students, 200),
    "Region": np.random.choice(regions, 200),
    "Gender": np.random.choice(genders, 200),
    "Subject": np.random.choice(subjects, 200),
    "Score": np.random.randint(50, 100, 200)
}
df = pd.DataFrame(data)

#Precompute all region-specific data
region_data = {}
for reg in ["All"] + regions:
    if reg == "All":
        dff = df
    else:
        dff = df[df["Region"] == reg]
    avg_subject = dff.groupby("Subject", as_index=False)["Score"].mean()
    avg_gender = dff.groupby("Gender", as_index=False)["Score"].mean()
    region_data[reg] = (avg_subject, avg_gender)

fig = make_subplots(rows=1, cols=2,
                    subplot_titles=("Average Score by Subject", "Average Score by Gender"),
                    specs=[[{"type": "bar"}, {"type": "pie"}]])

#Add traces for every region (so dropdown can toggle visibility)
visibility = []
for i, reg in enumerate(["All"] + regions):
    subj, gen = region_data[reg]
    # Bar chart (subjects)
    fig.add_trace(go.Bar(
        x=subj["Subject"], y=subj["Score"],
        name=f"{reg} - Subject Avg",
        visible=True if reg == "All" else False,
        marker_color='dodgerblue'
    ), row=1, col=1)

    # Pie chart (gender)
    fig.add_trace(go.Pie(
        labels=gen["Gender"], values=gen["Score"],
        name=f"{reg} - Gender Avg",
        hole=0.4,
        visible=True if reg == "All" else False
    ), row=1, col=2)

#Create dropdown buttons that control visibility
buttons = []
for i, reg in enumerate(["All"] + regions):
    # Determine visibility pattern (2 traces per region)
    visible = [False] * (2 * (len(regions) + 1))
    visible[i * 2] = True       # bar
    visible[i * 2 + 1] = True   # pie

    buttons.append(dict(
        label=reg,
        method="update",
        args=[{"visible": visible},
              {"title": f"🎓 Student Performance Dashboard — {reg}"}]
    ))

#Layout settings
fig.update_layout(
    updatemenus=[{
        "buttons": buttons,
        "direction": "down",
        "x": 0.45, "xanchor": "center",
        "y": 1.15, "yanchor": "top"
    }],
    title="Student Performance Dashboard — All Regions",
    height=650,
    width=950,
    template="plotly_white",
    showlegend=False
)

fig.show()
```

Our dashboard fits under the Real-Time data update type because it can refresh automatically to show new data as it changes. It includes user interaction through a dropdown filter that lets users view results for different regions, making it easier to focus on specific information. It serves an analytical function since it helps teachers or administrators study student performance patterns and identify which areas or subjects need improvement.

Create a mini dashboard project that tracks student grades or attendance:

```python
#Student Grades & Attendance Dashboard
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import plotly.express as px

#Simulated Data
np.random.seed(42)
students = [f"Student {i}" for i in range(1, 51)]
subjects = ["Math", "Science", "English"]
weeks = [f"Week {i}" for i in range(1, 9)]

data = {
    "Student": np.random.choice(students, 200),
    "Subject": np.random.choice(subjects, 200),
    "Week": np.random.choice(weeks, 200),
    "Grade": np.random.randint(60, 100, 200),
    "Attendance": np.random.randint(70, 100, 200)
}
df = pd.DataFrame(data)

# Compute Metrics
avg_grade_by_subject = df.groupby("Subject", as_index=False)["Grade"].mean()
avg_attendance_by_week = df.groupby("Week", as_index=False)["Attendance"].mean()
avg_grade_by_week = df.groupby("Week", as_index=False)["Grade"].mean()

#Interactive Feature (Dropdown by Subject)
fig = make_subplots(rows=2, cols=2,
                    subplot_titles=("Average Grade by Subject",
                                    "Attendance Rate by Week",
                                    "Grade Trend by Week",
                                    ""),
                    specs=[[{"type": "bar"}, {"type": "bar"}],
                           [{"colspan": 2}, None]])

# Chart 1 - Average Grade by Subject
fig.add_trace(go.Bar(x=avg_grade_by_subject["Subject"],
                     y=avg_grade_by_subject["Grade"],
                     name="Average Grade",
                     marker_color="dodgerblue"),
              row=1, col=1)

# Chart 2 - Attendance by Week
fig.add_trace(go.Bar(x=avg_attendance_by_week["Week"],
                     y=avg_attendance_by_week["Attendance"],
                     name="Attendance",
                     marker_color="mediumseagreen"),
              row=1, col=2)

# Chart 3 - Grade Trend by Week
for subj in subjects:
    dff = df[df["Subject"] == subj].groupby("Week", as_index=False)["Grade"].mean()
    fig.add_trace(go.Scatter(x=dff["Week"], y=dff["Grade"], mode="lines+markers",
                             name=subj), row=2, col=1)

# Dropdown to select a specific subject
dropdown_buttons = []
for subj in subjects:
    visible = [True, True] + [False] * len(subjects)
    # show the selected subject line only
    for i, s in enumerate(subjects):
        if s == subj:
            visible[2 + i] = True
    dropdown_buttons.append({
        "label": subj,
        "method": "update",
        "args": [{"visible": visible},
                 {"title": f"🎓 Student Dashboard – {subj} Focus"}]
    })

# Add an All Subjects button
visible = [True, True] + [True] * len(subjects)
dropdown_buttons.insert(0, {
    "label": "All Subjects",
    "method": "update",
    "args": [{"visible": visible},
             {"title": "Student Dashboard – All Subjects"}]
})

# Layout
fig.update_layout(
    updatemenus=[{
        "buttons": dropdown_buttons,
        "direction": "down",
        "x": 0.45, "xanchor": "center",
        "y": 1.15, "yanchor": "top"
    }],
    title="Student Dashboard – All Subjects",
    height=750, width=950,
    template="plotly_white",
    showlegend=True
)

fig.show()
```

This dashboard helps teachers easily see how students are doing in school. It shows the average grades for each subject, attendance for every week, and how grades change over time. The dropdown button lets teachers choose a subject to focus on. By looking at this dashboard, teachers can find which subjects students do well or struggle in and use that information to improve their lessons or help students who need support.
